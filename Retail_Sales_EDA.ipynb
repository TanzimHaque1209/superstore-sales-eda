```python
category_sales = df.groupby('Category')['Sales'].sum().reset_index()
```

```python
category_sales
```

```python
sns.barplot(x='Category', y='Sales', data=category_sales)

plt.title("Sales by Category")

plt.xlabel("Category")

plt.ylabel("Total Sales")

plt.show()
```

***The Technology category generated the highest total sales among all product categories, indicating strong customer demand and revenue contribution. Furniture also contributed significantly to sales, while Office Supplies recorded comparatively lower overall sales performance.***

```python
region_profit = df.groupby('Region')['Profit'].sum().reset_index()
region_profit
```

```python
sns.barplot(x='Region', y='Profit', data=region_profit)

plt.title("Profit by Region")

plt.xlabel("Region")

plt.ylabel("Total Profit")

plt.show()
```

***The West region generated the highest overall profit, suggesting strong business performance and profitability in that region. In comparison, other regions produced lower profits, indicating differences in regional sales efficiency and market performance.***

This exploratory data analysis project examined retail superstore sales data using Python libraries such as Pandas, Matplotlib, and Seaborn. The dataset was analyzed to identify sales trends, profitability patterns, regional performance, and category-wise business insights.

The analysis revealed that the Technology category achieved the highest sales performance, while the West region generated the highest profit. Correlation analysis also indicated that higher discount levels negatively impacted profitability. Additionally, several products generated losses despite maintaining sales activity, highlighting potential areas for business improvement.

Overall, the project demonstrated how exploratory data analysis can help uncover meaningful insights and support data-driven decision-making in retail business environments.

```python
monthly_sales = df.groupby('month')['Sales'].sum().reset_index()

plt.figure(figsize=(10,5))
sns.lineplot(x='month', y='Sales', data=monthly_sales, marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

***Sales peaked during November, indicating strong seasonal demand.***

```python
subcat_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=True)

plt.figure(figsize=(10,7))
subcat_sales.plot(kind='barh')
plt.title("Sales by Sub-Category")
plt.xlabel("Total Sales")
plt.ylabel("Sub-Category")
plt.show()
```

***Chairs and Phones were among the strongest-performing sub-categories by sales.***

```python
category_profit = df.groupby('Category')['Profit'].sum().reset_index()

plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Profit', data=category_profit)
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.show()
```

***Technology generated the highest profit among all categories.***

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Discount', y='Profit', data=df)
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```

***Higher discounts generally corresponded with lower profitability.***

```python
corr = df[['Sales', 'Profit', 'Discount', 'Quantity']].corr()

plt.figure(figsize=(7,5))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

***Profit showed a negative relationship with Discount, while Sales had a moderate positive relationship with Quantity.***

```python
segment_sales = df.groupby('Segment')['Sales'].sum().reset_index()

plt.figure(figsize=(8,5))
sns.barplot(x='Segment', y='Sales', data=segment_sales)
plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Total Sales")
plt.show()
```

***The Consumer segment generated the highest total sales.***
